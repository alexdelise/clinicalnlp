# 1: Classification

```python
from transformers import pipeline
classifier = pipeline(
    "text-classification",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)


texts = [
    "Patient reports chest pain and shortness of breath.",
    "Routine follow-up visit for blood pressure check.",
    "Mild headache for two days, no other symptoms."
]
for t in texts:
    print(t, "->", classifier(t))
```

# 2: Results

All three notes were classified as having negative sentiment. Given that all of these are negative medical symptoms, it is not necessarily incorrect, but it would certainly not be useful for a clinical setting, as they give no information at all about the actual symptoms. 

This model was trained on the Stanford Sentiment Treebank 2 (sst2), which is a database of 215,524 phrases parsed from 11,855 sentences from movie reviews. They were annotated for sentiment by three humans. 

This model is definitely not suitable for use in a hospital.
